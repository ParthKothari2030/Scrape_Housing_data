```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Reading source
f = open("Ahemdabad_source.txt","r",encoding="utf-8")
page_source = f.read()
f.close()
```

```python
# parsing the soup using html parser
soup = BeautifulSoup(page_source,'html.parser')
```

```python
project_divs = soup.find_all('div', class_='spc-img')
# scraped_data= soup.find_all('div',class_='spc-info')
```

```python
B = project_divs[1].find_next_sibling('div',class_='spc-info')
B
```

```python
# Finding the project name
scraped_projectName = soup.find_all('h2',class_='spc-title')

# print(scraped_projectName)
# Creating an array for Project name
Project_name = []
for projectname in scraped_projectName:
    Project_name.append(projectname.get_text().strip())

len(Project_name)
```

```python
# Finding the Contractor
scraped_Contractor = soup.find_all('p' ,class_="spc-location")

Contractor = []

for contractor in scraped_Contractor:
    Contractor.append(contractor.get_text().strip())
```

```python
Location = []
Group_name = []
for item in Contractor:
    parts = item.split('By')  # Split by 'By' to separate locality and group
    locality = parts[0].strip().replace("\n", "").replace("Ahmedabad", "").replace(",", "").strip()  
    group_name = parts[1].strip() if len(parts) > 1 else ""
    Location.append(locality)
    Group_name.append(group_name)
```

```python
# This get's you the names and location
# Beware there are locations with no contractores beacuse most probably they are plots
My_location = []
My_Group_Name= []
for i in range(len(scraped_Contractor)):
    test = scraped_Contractor[i].get_text().strip().split('By')
    # print(test)
    locality = test[0].replace("Ahmedabad", "").replace(",", "").strip()
    My_location.append(locality)
    if len(test) > 1:
        My_Group_Name.append(test[1].strip())
    else:
        My_Group_Name.append(None)

```
